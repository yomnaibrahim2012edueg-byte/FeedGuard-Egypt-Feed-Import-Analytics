```python
yearly_sums = df[mask_years].groupby('refYear')[['netWgt', 'cifvalue']].sum()
yearly_sums['imputation_ratio'] = yearly_sums['netWgt'] / yearly_sums['cifvalue']

yearly_sums
```

# Compensation for lost weight

```python
AvgKg_2018 = yearly_sums.loc[2018, 'imputation_ratio']
AvgKg_2019 = yearly_sums.loc[2019, 'imputation_ratio']
```

```python
AvgKg_2018
```

```python
AvgKg_2019
```

```python

mask_2018 = (df['refYear'] == 2018) & df['netWgt'].isna()

df.loc[mask_2018, 'netWgt'] = df.loc[mask_2018, 'cifvalue'] * AvgKg_2018


mask_2019 = (df['refYear'] == 2019) & df['netWgt'].isna()

df.loc[mask_2019, 'netWgt'] = df.loc[mask_2019, 'cifvalue'] * AvgKg_2019
```

```python
df.isnull().sum()
```

# Identifying the years that contain missing values

```python
df[df['netWgt'].isna()]['refYear'].value_counts()
```

```python
yearly_ratios = df.groupby('refYear')['netWgt'].sum() / df.groupby('refYear')['cifvalue'].sum()

missing_mask = df['netWgt'].isna()
df.loc[missing_mask, 'netWgt'] = (
    df.loc[missing_mask, 'cifvalue'] * df.loc[missing_mask, 'refYear'].map(yearly_ratios)
)
```

# Check if there are zeros in cifvalue or netWgt 

```python
print("عدد الأصفار في netWgt:", (df['netWgt'] == 0).sum())
print("عدد الأصفار في cifvalue:", (df['cifvalue'] == 0).sum())
```

```python
zero_mask = (df['netWgt'] == 0)

df.loc[zero_mask, 'netWgt'] = (
    df.loc[zero_mask, 'cifvalue'] * df.loc[zero_mask, 'refYear'].map(yearly_ratios)
)
```
